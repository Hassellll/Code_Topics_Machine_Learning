```python
#We import libraries

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
#Load the data

data = pd.read_csv("Stochastic_Descent_(SGD).csv")
```

```python
# Display the first rows of the dataset

data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
#Variables

X = data[["x1","x2"]]
Y = data["y"]
```

```python
#Train / Test

x_train,x_test,y_train,y_test = train_test_split(
    X,Y, test_size=0.2, random_state=42
)
```

```python
# Feature Scaling

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#SGD Model

model = SGDRegressor(
    max_iter=1000,
    learning_rate="optimal",
    eta0=0.01,
    random_state=42
)
```

```python
#Training

model.fit(x_train, y_train)
```

```python
#Prediction

y_pred = model.predict(x_test)
```

```python
#Evaluation

mse = mean_squared_error(y_test, y_pred)
print("MSE",mse)
```

```python
# Visualization

plt.figure(figsize=(7,5))

#Actual vs. Forecast Points
plt.scatter(y_test,y_pred, color="blue",label="Predictions")


```

```python
# Ideal straight line (y = x)

min_val = min(min(y_test), min(y_pred))
max_val = max(max(y_test),max(y_pred))
plt.plot([min_val,max_val],[min_val, max_val], "r--", label="ideal (y=x)")

plt.xlabel("real values")
plt.ylabel("predicted values")
plt.title("SGD Regression: Actual vs. Predicted")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#Result:

#The graph compares the actual values with the values predicted by the SGD model
#The red dashed line represents the ideal prediction (y = x). The closer the points are to this line, the more accurate the model's predictions and the lower the prediction error.
```
