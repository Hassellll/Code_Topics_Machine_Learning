```python
#We import the libraries

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
#Load the dataset

data = pd.read_csv("Random_Forest_Clustering.csv")
```

```python
#We see our data in a table

data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
# Features and target
X = data.drop("target", axis=1)
y = data["target"]
```

```python
# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Train Random Forest
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_scaled, y)
```

```python
# Obtain leaf nodes
leaf_nodes = rf.apply(X_scaled)
```

```python
# Cluster leaf representations
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

data["cluster"] = kmeans.fit_predict(leaf_nodes)

print(data.head())
```
