```python
#Import Pandas and make it read the CSV data



import pandas as pd

data = pd.read_csv("One_hot_Label_Target_Kfold.csv")

print(data)
```

```python
#Label encoding converts text to a number


from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data["city_label"] = le.fit_transform(data["city"]) 

print(data)

```

```python
#One Hot Encoding
#Convert a category into multiple columns

data_onehot = pd.get_dummies(data,columns=["city"])

#pd.get_dummies()
#Creates binary variables.

#columns=[“City”]

#It tells you:
#“Convert the City column”

print(data_onehot)
```

```python
#Target Encoding

#Use the target mean
#Target is the variable we want to predict


import category_encoders as ce

x = data[["city", "gender", "age"]]
y = data["purchase"]
```

```python
target_encoder = ce.TargetEncoder(cols=["city"])
#import the special encoding library

x_target = target_encoder.fit_transform(x,y)
#Selecting input variables

print(x_target)
#We select the target audience
```

```python
#K-Fold Target Encoding

#Split the data into parts

from sklearn.model_selection import KFold
import numpy as np

data["city_kfold_target"] = np.nan

kf = KFold(n_splits=3, shuffle=True, random_state=42)

for train_idx, val_idx in kf.split(data):

    train_df = data.iloc[train_idx]
    val_df = data.iloc[val_idx]

    means = train_df.groupby("city")["purchase"].mean()

    data.loc[val_idx, "City_KFold_Target"] = val_df["city"].map(means)

print(data)
```
