```python
#We import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Load the dataset

data = pd.read_csv("Gradient_Descent.csv")
```

```python
#We see our data 

data.head()
```

```python
data.describe()
```

```python
#We use columns so we don't get confused when defining variables

data.columns
```

```python
X = data["Study_Hours"].values
Y = data["Score"].values
```

```python
#Initialize parameters

w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 500
```

```python
n = len(X)

loss_history = []
```

```python
#Training

for i in range(epochs):
    #Prediction
    y_pred = w * X + b

    #Error

    error = y_pred - Y

    #Gradients

    dw = (2/n) * np.sum(error * X)
    db = (2/n) * np.sum(error)

    #Update

    w -= learning_rate * dw
    b -= learning_rate * db

    #Save loss

    loss = np.mean(error**2)
    loss_history.append(loss)
```

```python
#Final Results:

print(f"Modelo final: y = {w:.2f}X + {b:.2f}")
```

```python
#Graph 1 
plt.figure(figsize=(6,4))
plt.scatter(X, Y, color="blue", label="Real data")

plt.plot(X, w*X + b, color="red", label="Learned Model")

plt.title("Gradient Descent Regression")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.legend()
plt.grid()
plt.show()
```

```python
#Graph 1

#Shows what the model learned.

#Question:

#Does the line accurately represent the data?
```

```python
#GRÁFICA 2: pérdida (loss)

plt.figure(figsize=(6,4))
plt.plot(loss_history, color="purple")

plt.title("Evolution of the error (Loss)")
plt.xlabel("Epochs")
plt.ylabel("MSE")
plt.grid()
plt.show()
```
