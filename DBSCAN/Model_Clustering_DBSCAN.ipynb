```python
#We import the libraries:

import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.cluster import DBSCAN
```

```python
#We load the dataset:

data =pd.read_csv("DBSCAN_Clustering.csv")
```

```python
#Visualization

data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
#Select features

X = data[["X","Y"]]
```

```python
#Apply DBSCAN

model = DBSCAN(eps=0.5, min_samples=3)
data["Cluster"] = model.fit_predict(X)
```

```python
#Display the clusters

print(data)
```

```python
#Plot the clusters

plt.figure(figsize=(8,6))
plt.scatter(data["X"], data["Y"], c = data["Cluster"], cmap="viridis",s=80)

#Plot the clusters

plt.figure(figsize = (6,6))
plt.scatter(data["X"],data["Y"],c=data["Cluster"], cmap="viridis",s =80)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("DBSCAN Clustering")
plt.grid(True)
plt.show
```
