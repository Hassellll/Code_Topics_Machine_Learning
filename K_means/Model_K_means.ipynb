```python
#Import libraries

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#Load the CSV dataset

data = pd.read_csv("K-means.csv")
```

```python
#Select the features

X = data[["Age","Income"]]
```

```python
#Create the K-Means model with 3 clusters

model = KMeans(n_clusters=3, random_state=42)
```

```python
#Train the model

model.fit(X)
```

```python
#Assign each record to a cluster

data["Cluster"] = model.labels_
```

```python
#Display the results

print(data)
```

```python
#Get the cluster centroids

centroids = model.cluster_centers_
```

```python
#Plot the clusters

plt.figure(figsize=(8,6))

plt.scatter(
    data['Age'],
    data['Income'],
    c=data['Cluster'],
    cmap='viridis',
    s=80
)
```

```python
#Plot the centroids

plt.scatter(
    centroids[:,0],
    centroids[:,1],
    color="red",
    marker="X",
    s=250,
    label="Centroids"
)

plt.xlabel("Age")
plt.ylabel("Income")
plt.title("K-Means Clustering")
plt.legend()
plt.show()
```
