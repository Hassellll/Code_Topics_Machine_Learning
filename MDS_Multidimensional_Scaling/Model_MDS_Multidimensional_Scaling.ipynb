```python
#Code with the topic : MDS Muldimensional Scaling

#Objective:

#Reduce multidimensional data into a two-dimensional representation
#while preserving pairwise distances as much as possible.

#Technologies:

#Pandas
#Scikit learn
#Jupyter notebook
#Visual Studio Code

#Method used:

#MDS
```

```python
#We import the libraries

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.manifold import MDS
```

```python
# load the dataset CSV

data = pd.read_csv("MDS_Multidimensional_Scaling.csv")
```

```python
#Separate features and labels

x = data.iloc[:,1:] #Numeric Columns
labels= data.iloc[:, 0] #Student names
```

```python
#Apply Muldimensional Scaling (MDS)

mds = MDS(
    n_components=2,
    random_state=42,
    dissimilarity="euclidean"
)

x_mds = mds.fit_transform(x)
```

```python
#Create a Dataframe with the transformed coordinates

mds_df = pd.DataFrame(x_mds, columns = ["MDS1","MDS2"])
mds_df["student"] = labels

print(mds_df)
```

```python
#Plot the results 

plt.figure(figsize=(8,6))
plt.scatter(mds_df["MDS1"], mds_df["MDS2"], color = "steelblue")
```

```python
#Add labels to each point 

for i , txt in enumerate(mds_df["student"]):
                         plt.annotate(txt,
                                      (mds_df["MDS1"][i],mds_df["MDS2"][i]),
                                      fontsize=9)


plt.title("Multidimensional Scaling (MDS)")
plt.xlabel("MDS Dimension 1 ")
plt.ylabel("MDS Dimension 2 ")
plt.grid(True)
plt.show()
```
